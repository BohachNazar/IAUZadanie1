#### Úprava formátu dátumov na jednotný formát:

```python
def normalize_date_series(s):
    s_clean = s.astype(str).str.strip()
    # odstránenie času
    s_clean = s_clean.str.replace(r",\s*\d{2}:\d{2}:\d{2}$", "", regex=True)
    # nahradenie bodiek a medzier
    s_clean = s_clean.str.replace(".", "/", regex=False)
    s_clean = s_clean.str.replace(r"\s+", " ", regex=True)
    
    #viacnásobná konverzia
    out = pd.to_datetime(s_clean, errors='coerce')
    mask = out.isna()
    out2 = pd.to_datetime(s_clean[mask], errors='coerce', dayfirst=True)
    out.loc[mask] = out2
    
    common_formats = [
        "%d %b %Y", 
        "%d %B %Y",      
        "%Y-%m-%d",      
        "%Y/%m/%d",      
        "%d/%m/%Y",      
        "%m/%d/%Y",      
    ]
    mask = out.isna()
    if mask.any():
        for fmt in common_formats:
            try:
                tmp = pd.to_datetime(s_clean[mask], format=fmt, errors='coerce')
                fill_mask = mask.copy()
                fill_mask.loc[:] = False
                fill_mask[mask] = tmp.notna().values
                out.loc[fill_mask] = tmp[tmp.notna()]
                mask = out.isna()
                if not mask.any():
                    break
            except Exception:
                pass
    return out

patient['registration_dt'] = normalize_date_series(patient['registration'])
station['revision_dt'] = normalize_date_series(station['revision'])

print('Neúspešné konverzie po normalizácii - patient:', patient['registration_dt'].isna().sum())
print('Neúspešné konverzie po normalizácii - station:', station['revision_dt'].isna().sum())
```

Vytvorili sme si pomocne stĺpce, kde máme upravené dátumy:

stlpec `registration_dt` v patient

stlpec `revision_dt` v station

(C-1b) Vychýlené hodnoty (outlier detection), vyskúšajte riešiť problém min. 2 technikami
- odstránenie vychýlených alebo odľahlých pozorovaní
- nahradenie vychýlenej hodnoty hraničnými hodnotami rozdelenia (napr. 5%, 95%)

1) IQR (InterQuartile Range) – odstránenie riadkov s outliermi.
2) Standard deviation (štandardná odchýlka) – 5. a 95. percentil

```python
# Detekcia a odstránenie IQR outlierov (bez geo stĺpcov)
geo_exclude = [c for c in ['latitude', 'longitude'] if c in obs.columns]
num_cols = [c for c in obs.select_dtypes(include=['number']).columns if c not in geo_exclude]
iqr_df = obs.copy()

row_is_outlier = pd.Series(False, index=iqr_df.index)
per_col_counts = {}
for col in num_cols:
    Q1 = iqr_df[col].quantile(0.25)
    Q3 = iqr_df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    mask = (iqr_df[col] < lower) | (iqr_df[col] > upper)
    per_col_counts[col] = int(mask.sum())
    row_is_outlier |= mask

print("Počty IQR outlierov na stĺpec (top 10):")
print(sorted(per_col_counts.items(), key=lambda x: x[1], reverse=True)[:10])
print(f"Riadky s outlierom aspoň v jednom stĺpci: {int(row_is_outlier.sum())}")
print("Tvar pôvodný:", obs.shape)
obs_iqr = iqr_df[~row_is_outlier].copy()
print("Tvar po odstránení IQR outlierov:", obs_iqr.shape)
```

```python
#Technika 2 Standard Deviation (winsorizácia bez geo stĺpcov)
sd_df = obs.copy()
geo_exclude = [c for c in ['latitude', 'longitude'] if c in sd_df.columns]
num_cols = [c for c in sd_df.select_dtypes(include=['number']).columns if c not in geo_exclude]
summary = []
for col in num_cols:
    col_series = sd_df[col]
    mean, std = col_series.mean(), col_series.std()
    if std == 0 or np.isnan(std):
        summary.append({'col': col, 'z_outliers': 0, 'p5': np.nan, 'p95': np.nan})
        continue
    z = np.abs((col_series - mean) / std)
    z_outliers = int((z > 3).sum())
    p5, p95 = col_series.quantile(0.05), col_series.quantile(0.95)
    sd_df[col] = col_series.clip(lower=p5, upper=p95)
    summary.append({'col': col, 'z_outliers': z_outliers, 'p5': p5, 'p95': p95})

print("Z-score outliery (top 10):")
print(sorted([(s['col'], s['z_outliers']) for s in summary], key=lambda x: x[1], reverse=True)[:10])
print("Tvar po winsorizácii:", sd_df.shape)
obs_sd = sd_df
```

# 1.3 Formulácia a štatistické overenie hypotéz o dátach
**Hypotézy:**
- $H_0$: Priemerná srdcová frekvencia (HR) je rovnaká v skupinách s oximetry=1 a oximetry=0.
- $H_1$: Priemerná srdcová frekvencia (HR) sa líši medzi skupinami s oximetry=1 a oximetry=0.

## Príprava skupín

```python
# Skupiny podľa oximetry z vyčistených dát
hr_ox0 = obs_iqr.loc[obs_iqr['oximetry'] == 0, 'HR'].dropna()
hr_ox1 = obs_iqr.loc[obs_iqr['oximetry'] == 1, 'HR'].dropna()
print(f'Počet pozorovaní oximetry=0: {len(hr_ox0)}')
print(f'Počet pozorovaní oximetry=1: {len(hr_ox1)}')
```

```python
# Shapiro-Wilk test normality
import scipy.stats as stats
shap0 = stats.shapiro(hr_ox0.sample(min(5000, len(hr_ox0)), random_state=1) if len(hr_ox0) > 5000 else hr_ox0)
shap1 = stats.shapiro(hr_ox1.sample(min(5000, len(hr_ox1)), random_state=1) if len(hr_ox1) > 5000 else hr_ox1)
print(f'oximetry=0 : p-value={shap0.pvalue:.4f}')
print(f'oximetry=1 : p-value={shap1.pvalue:.4f}')
```

```python
# Vizualizácia distribúcie - histogramy
sns.histplot(hr_ox0, label='oximetry=0', alpha=0.6, bins=30)
sns.histplot(hr_ox1, label='oximetry=1', alpha=0.6, bins=30)
plt.xlabel('HR')
plt.ylabel('Počet')
plt.title('Distribúcia HR podľa oximetry (po odstránení outlierov)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

Z výstupu Shapiro-Wilk testu vidíme, že p-hodnota je menšia ako 0.05 aspoň v jednej skupine. To znamená, že aspoň jedna skupina nepochádza z normálneho rozdelenia. Preto použijeme Mann-Whitney U test.

```python
# Shapiro-Wilk test
shap0 = stats.shapiro(hr_ox0.sample(min(5000, len(hr_ox0)), random_state=1) if len(hr_ox0) > 5000 else hr_ox0)
shap1 = stats.shapiro(hr_ox1.sample(min(5000, len(hr_ox1)), random_state=1) if len(hr_ox1) > 5000 else hr_ox1)

print(f"oximetry=0 : p-value={shap0.pvalue:.4f}")
print(f"oximetry=1 : p-value={shap1.pvalue:.4f}")

if shap0.pvalue < 0.05:
    print("oximetry=0 nepochádza z normálneho rozdelenia (p < 0.05)")
else:
    print("oximetry=0 pochádza z normálneho rozdelenia (p ≥ 0.05)")
    
if shap1.pvalue < 0.05:
    print("oximetry=1 nepochádza z normálneho rozdelenia (p < 0.05)")
else:
    print("oximetry=1 pochádza z normálneho rozdelenia (p ≥ 0.05)")
```

## Podobnosť variancií

**Leveneho test** testuje $H_0$: všetky skupiny majú rovnaké variancie. Ak nezamietame $H_0$ ($p > 0.05$), variancie sú podobné.

```python
# Leveneho test na rovnosť variancií
levene = stats.levene(hr_ox0, hr_ox1)
print(f'Levene test: statistic={levene.statistic:.4f}, p-value={levene.pvalue:.4f}')
```

Z výstupu Leveneho testu vidíme, že p-hodnota je menšia ako 0.05, takže variancie sa štatisticky významne líšia.

```python
# Mann-Whitney U test
mannwhitney = stats.mannwhitneyu(hr_ox0, hr_ox1, alternative='two-sided')
print(f'Mann-Whitney U test: p-value={mannwhitney.pvalue:.4g}')
```

Z výstupu Mann-Whitney U testu vidíme, že p-hodnota je menšia ako 0.05, takže zamietame nulovú hypotézu. Priemerná HR sa štatisticky významne líši medzi skupinami oximetry=0 a oximetry=1.

```python
# Confidence intervaly
import statsmodels.stats.api as sms
ci_ox0 = sms.DescrStatsW(hr_ox0).tconfint_mean()
ci_ox1 = sms.DescrStatsW(hr_ox1).tconfint_mean()
print(f'95% Confidence interval pre oximetry=0: [{ci_ox0[0]:.2f}, {ci_ox0[1]:.2f}]')
print(f'95% Confidence interval pre oximetry=1: [{ci_ox1[0]:.2f}, {ci_ox1[1]:.2f}]')
```

```python
# Barplot s confidence intervalmi
import seaborn as sns
import matplotlib.pyplot as plt
comparison_data = obs_iqr[obs_iqr['oximetry'].isin([0, 1])].copy()
sns.barplot(x='oximetry', y='HR', data=comparison_data, capsize=0.1, err_kws={'linewidth': 2})
plt.xlabel('Oximetry')
plt.ylabel('HR (priemer)')
plt.title('Porovnanie priemernej HR podľa oximetry\n(s 95% confidence intervalmi)')
plt.xticks([0, 1], ['oximetry=0', 'oximetry=1'])
plt.grid(True, alpha=0.3, axis='y')
plt.show()
```

Z confidence intervalov a grafu vidíme, že skupina s oximetry=1 má vyššiu priemernú HR a intervaly sa neprekrývajú. To potvrdzuje štatisticky významný rozdiel.

```python
plt.show()
from statsmodels.stats.power import TTestIndPower

# veľkosti vzoriek v oboch skupinách
n1 = len(hr_ox0)
n2 = len(hr_ox1)

# veľkosť efektu (Cohen's d) - hrubý odhad
mean_diff = hr_ox0.mean() - hr_ox1.mean()
pooled_std = ((hr_ox0.std() ** 2 + hr_ox1.std() ** 2) / 2) ** 0.5
effect_size = mean_diff / pooled_std

# výpočet sily testu
analysis = TTestIndPower()
power = analysis.solve_power(effect_size=abs(effect_size), nobs1=min(n1, n2), alpha=0.05)

print(f"Veľkosť efektu (Cohen's d): {effect_size:.3f}")
print(f"Štatistická sila testu: {power:.3f}")
```

Výsledky analýzy štatistickej sily ukazujú, že test mal dostatočnú podporu z dát.
Odhadnutá veľkosť efektu (Cohen’s d = -0.587) naznačuje veľmi silný efekt medzi skupinami s oximetriou a bez nej.
Štatistická sila testu (power = 1.00) potvrdzuje, že použité testy mali dostatočnú schopnosť odhaliť rozdiel medzi skupinami, teda naše závery sú spoľahlivé.

- $H_1$: Priemerná hodnota Hb je nižšia v skupine s oximetriou (oximetry=1).

### (A-1b) Train/Test Split

### (B-1b) Transformácia na numerický formát

Transformujte dáta na vhodný formát pre strojové učenie:
- Jedno pozorovanie = jeden riadok 
- Každý atribút musí byť v numerickom formáte
- Iteratívne integrujte kroky predspracovania z prvej fázy

```python
obj_cols = x_train.select_dtypes(include=['object']).columns.tolist()
print('Object stĺpce (textové kategórie):', obj_cols, '\n')
```

#### Krok: kódovanie `code` a `QoS` na čísla
- `code`, `QoS` sú typu object → prevedieme ich na numerické pomocou OneHotEncoder
- `station` (vysoká kardinalita) a `revision` (textový dátum) odstranime

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

geo_cols = [c for c in ['latitude','longitude'] if c in x_train.columns]
num_cols = [c for c in x_train.select_dtypes(include=['number']).columns if c not in geo_cols]
obj_cols = [c for c in ['code','QoS'] if c in x_train.columns]
cols_to_drop = [c for c in ['station','revision'] if c in x_train.columns]

onehot_b = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

preproc_basic = ColumnTransformer(
    transformers=[
        ('num', 'passthrough', num_cols),
        ('cat', onehot_b, obj_cols),
        ('geo', 'passthrough', geo_cols),
        ('drop', 'drop', cols_to_drop),
    ],
    remainder='drop'
)

x_train_final = preproc_basic.fit_transform(x_train)
x_test_final  = preproc_basic.transform(x_test)

print('numeric dtype:', np.issubdtype(x_train_final.dtype, np.number))
```

### (C-1b) Preprocessing pipelines: Scaling + Transformers 
Aplikujeme len numerické transformácie:
- 2 scaling techniky (StandardScaler, MinMaxScaler)
- 2 transformers (PowerTransformer Yeo-Johnson, QuantileTransformer)

```python
# (C-1b) 4 numerické transformácie na x_train_final/x_test_final
from sklearn.preprocessing import StandardScaler, MinMaxScaler, PowerTransformer, QuantileTransformer

print('Vstupné tvary ->', 'x_train_final:', x_train_final.shape, '| x_test_final:', x_test_final.shape)

# 1) StandardScaler
scaler_std = StandardScaler()
Xt_std = scaler_std.fit_transform(x_train_final)
Xv_std = scaler_std.transform(x_test_final)
print('StandardScaler ->', Xt_std.shape, Xv_std.shape)

# 2) MinMaxScaler
scaler_mm = MinMaxScaler()
Xt_mm = scaler_mm.fit_transform(x_train_final)
Xv_mm = scaler_mm.transform(x_test_final)
print('MinMaxScaler ->', Xt_mm.shape, Xv_mm.shape)

# 3) PowerTransformer (Yeo-Johnson)
pt = PowerTransformer(method='yeo-johnson', standardize=True)
Xt_pt = pt.fit_transform(x_train_final)
Xv_pt = pt.transform(x_test_final)
print('PowerTransformer ->', Xt_pt.shape, Xv_pt.shape)

# 4) QuantileTransformer (normal)
qt = QuantileTransformer(output_distribution='normal', n_quantiles=max(10, int(np.sqrt(len(x_train_final)))))
Xt_qt = qt.fit_transform(x_train_final)
Xv_qt = qt.transform(x_test_final)
print('QuantileTransformer ->', Xt_qt.shape, Xv_qt.shape)
```

## Jednoduché vizuálne porovnanie transformácií (HR)

Ukážka pre jeden číselný atribút (HR):
- 2×2 histogram: pôvodné dáta vs. Standard, MinMax, Power (Yeo–Johnson)

3.1 Jednoduchý klasifikátor na základe závislosti v dátach  (5b)

(A-2b) Naimplementujte jednoduchý ID3 klasifikátor s hĺbkou min 2 (vrátane root/koreň). 

```python
# vyberie sa jeden koreňový atribút s najvyšším information gain
# pre každú vetvu sa predikuje väčšinová trieda

import pandas as pd

class ID3Depth2:
    def __init__(self, n_bins=3):
        self.n_bins = n_bins
        self.root_feature = None
        self.is_numeric = False
        self.bin_edges = None
        self.leaf_prediction = {}
        self.majority_class = None

    def _entropy(self, y):
        p = y.value_counts(normalize=True)
        return -(p * np.log2(p + 1e-12)).sum()

    def _info_gain_categorical(self, y, x):
        H = self._entropy(y)
        parts = x.fillna('MISSING').astype(str)
        weighted = 0.0
        for val, idx in parts.groupby(parts, observed=True).groups.items():
            y_sub = y.loc[idx]
            weighted += (len(y_sub) / len(y)) * self._entropy(y_sub)
        return H - weighted

    def _info_gain_numeric(self, y, x):
        H = self._entropy(y)
        try:
            binned, bins = pd.qcut(x, q=self.n_bins, duplicates='drop', retbins=True)
        except ValueError:
            return -np.inf, None
        weighted = 0.0
        for val, idx in binned.groupby(binned, observed=True).groups.items():
            y_sub = y.loc[idx]
            weighted += (len(y_sub) / len(y)) * self._entropy(y_sub)
        return H - weighted, bins

    def fit(self, X, y):
        self.majority_class = y.mode().iloc[0]
        best_ig = -np.inf
        best_feature = None
        best_is_numeric = False
        best_bins = None

        for col in X.columns:
            series = X[col]
            if pd.api.types.is_numeric_dtype(series):
                ig, bins = self._info_gain_numeric(y, series)
                if ig > best_ig:
                    best_ig = ig
                    best_feature = col
                    best_is_numeric = True
                    best_bins = bins
            else:
                ig = self._info_gain_categorical(y, series)
                if ig > best_ig:
                    best_ig = ig
                    best_feature = col
                    best_is_numeric = False
                    best_bins = None

        self.root_feature = best_feature
        self.is_numeric = best_is_numeric
        self.bin_edges = best_bins

        #vytvorenie listov (predikcií) pre jednotlivé vetvy
        if self.is_numeric:
            binned = pd.cut(X[self.root_feature], bins=self.bin_edges, include_lowest=True, duplicates='drop')
            for interval, idx in binned.groupby(binned, observed=True).groups.items():
                y_sub = y.loc[idx]
                self.leaf_prediction[interval] = y_sub.mode().iloc[0]
        else:
            parts = X[self.root_feature].fillna('MISSING').astype(str)
            for val, idx in parts.groupby(parts, observed=True).groups.items():
                y_sub = y.loc[idx]
                self.leaf_prediction[val] = y_sub.mode().iloc[0]

        #informácie o naučenom strome (koreň + listy)
        print("ID3 (h=2) natrénovaný.")
        print("Koreňový atribút:", self.root_feature)
        if self.is_numeric:
            print("Typ atribútu: numerický, počet košov:", len(self.leaf_prediction))
            for interval, pred in sorted(self.leaf_prediction.items(), key=lambda kv: kv[0].left):
                print(" ", interval, "-> predikovaná trieda:", pred)
        else:
            print("Typ atribútu: kategóriálny, počet hodnôt:", len(self.leaf_prediction))
            for val, pred in self.leaf_prediction.items():
                print(" ", val, "-> predikovaná trieda:", pred)

    def predict(self, X):
        preds = []
        if self.root_feature is None:
            return np.full(len(X), self.majority_class)
        if self.is_numeric:
            binned = pd.cut(X[self.root_feature], bins=self.bin_edges, include_lowest=True, duplicates='drop')
            for val in binned:
                pred = self.leaf_prediction.get(val, self.majority_class)
                preds.append(pred)
        else:
            parts = X[self.root_feature].fillna('MISSING').astype(str)
            for val in parts:
                pred = self.leaf_prediction.get(val, self.majority_class)
                preds.append(pred)
        return np.array(preds)

#tréning na už pripravených dátach (X_train, y_train)
id3 = ID3Depth2(n_bins=3)
id3.fit(X_train, y_train)
```

```python
# 3.1 (B) Vyhodnotenie ID3 klasifikátora: accuracy, precision, recall
from sklearn.metrics import accuracy_score, precision_score, recall_score, classification_report

y_pred_id3 = id3.predict(X_test)

print("Accuracy:", round(accuracy_score(y_test, y_pred_id3), 3))
print("Precision:", round(precision_score(y_test, y_pred_id3), 3))
print("Recall:", round(recall_score(y_test, y_pred_id3), 3))
print("\nClassification report:\n", classification_report(y_test, y_pred_id3))
```

### 3.1 (B) Zhrnutie metrík ID3

- Accuracy: 0.755 – približne tri štvrtiny predikcií sú správne.
- Precision: 0.731 – keď model predpovie triedu 1, v ~73 % prípadov má pravdu.
- Recall: 0.886 – model zachytí väčšinu pozitívnych prípadov (má vysokú citlivosť na triedu 1).

Podrobnosti z reportu:

### 3.1 (C) Overfitting – záver

- ID3 (n_bins=3): Train vs. Test sú veľmi podobné (Acc 0.761 vs 0.755, Prec 0.748 vs 0.731, Rec 0.874 vs 0.886) – nepozorujeme overfitting.
- ID3 (n_bins=2): Train ≈ Test (Acc 0.798 vs 0.797, Prec 0.868 vs 0.857, Rec 0.761 vs 0.761) – veľmi stabilné, menšia modelová zložitosť.

Trade-off:
- n_bins=3 má vyšší recall pre triedu 1 (0.886), ale nižšiu accuracy a precision.
- n_bins=2 má vyššiu accuracy a precision, no nižší recall.

Rozhodnutie:
- Ak je prioritou zachytiť čo najviac pozitívnych prípadov (recall), je vhodné n_bins=3.
- Pre vyváženejší a stabilný model bez overfitu budeme ďalej používať ID3 s n_bins=2.
